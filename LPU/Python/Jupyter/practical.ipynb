Car laps timing

```python
n = 10
laps=[]

for x in range(n):
    s = x+1
    laps_time= int(input(f"Enter the timing for lap {s}"))
    laps.append(laps_time)
s = 1
for x in laps:
    print(f"Timing for lap {s} is {x} sec.")
    s = s+1

print(f"Best lap timing is {min(laps)}")
```

Sum of natural numbers using recursion.

```python
def naturalSum(x):
    if x == 1:
        return 1
    return x + naturalSum(x-1)

n = int(input("Enter the last natural no. to know the sum"))

print("Sum of natural no. is ", naturalSum(n))
```

Dictionary Problem 2

```python
n = int(input("Enter no. of students"))

marks = {}

for i in range(0,n):
    key = input("Enter the name of student.")
    value = []
    for i in range(5):
        value.append(int(input("Enter marks: ")))
    marks[key] = value

for key,value in marks.items():
    print("Marks of ", key, " is ", value)

highKey = ""
highVal = 0

for key, value in marks.items():
    if int(sum(value)) > highVal:
        highKey = key
        highVal = value

print("Highest score is ", marks[highVal]," of ", marks[highKey])
```

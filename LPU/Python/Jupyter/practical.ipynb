```python
balance = 200000
option = int(input("1.Withdraw \n2.Deposit\n3.Exit"))

while( balance > 0 and option !=3):
    if(option == 3):
        if(balance<=0):
            break
        print("Your balance is: ",balance)
        print("Exiting Now...")
        break
    if(option == 1):
        if(balance<=0):
            break
        withdraw = int(input("Enter the amount to withdraw.: "))
        balance = balance - withdraw
        print("You withdrew ", withdraw, "Rs.")
        print("Your balance is: ",balance)
        option = int(input("1.Withdraw \n2.Deposit\n3.Exit"))
    elif(option == 2):
        if(balance<=0):
            break
        deposit = int(input("Enter the amount to deposit.: "))
        balance = balance + deposit
        print("You Deposited ", deposit, "Rs.")
        print("Your balance is: ",balance)
        option = int(input("1.Withdraw \n2.Deposit\n3.Exit"))
        


```

```python
def calculate_bill():
    bill_amount = int(input("Enter the bill amount in Rs."))
    GST = bill_amount * (5/100)
    Total_amount = bill_amount + GST
    if(bill_amount >= 2000):
        Discount = (Total_amount*(10/100))
        Total_amount = Total_amount - Discount
    elif(bill_amount<2000):
        Discount = 0
    print("Your original bill is: ", bill_amount)
    print("GST amount is: ", GST)
    print("Your Discount is: ", Discount)
    print("Your total amount is: ", Total_amount)

calculate_bill()
```
